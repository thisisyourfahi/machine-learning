```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../module-1/titanic_data_updated.csv')
```

#### Multivariate Analysis

```python
# gender wise survival rate

sns.countplot(x=df['Sex'], hue=df['Survived'])
```

```python
gender_groupby = df.groupby('Sex')['Survived'].value_counts(normalize=True) * 100
gender_groupby
```

```python
# passenger class wise survival rate

sns.countplot(x=df['Pclass'], hue=df['Survived'])
```

```python
class_groupby = df.groupby('Pclass')['Survived'].value_counts(normalize=True) * 100
class_groupby
```

```python
sns.barplot(x=df['Pclass'], y=df['Fare'], errorbar=None)
```

```python
# kde plot

sns.kdeplot(x=df['Age'], hue=df['Survived'])
```
